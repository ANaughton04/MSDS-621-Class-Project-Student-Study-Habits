# Categorical Conversions to Numeric
We want to convert all of our categorical columns to numeric values so that we can use them for data analysis and machine learning.  The categorical values fall into two categories:

## Label Encoding
These data points aren't necessary numeric but preserve some sort of "ordering" to the values.  They can converted to simple 0, 1, 2, 3, 4 mappings.  An example for our dataset would be time slept.

## One-Hot Encoding
These data points don't have any ordering value to them.  We need to create a new column for each unique value in the column and include either a 1 or a 0 for that specific row.  An example for our dataset would be education.  This will increase our dimensionality but it is what we have to do for the time-being.

```python
df['exam_percentage'].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder

encoding_dict = {
    'year': ['first_year', 'second_year', 'third_year', 'fourth_year'],
    'study_hours': ['less_than_1_hour', '1-2_hours', '2-3_hours', '3-5_hours', 'more_than_5_hours'],
    'sleep_hours': ['less_than_5_hours', '5-6_hours', '6-7_hours', '7-8_hours', 'more_than_8_hours'],
    'attendance': ['below_50%', '50-65%', '66-75%', '76-90%', 'above_90%'],
    'revision': ['rarely', 'only_before_exams', 'once_a_week', 'several_times_a_week', 'daily'],        #Need to discuss if rarely or only before exams is actually the lowest value
    'exam_prep': ['no_fixed_routine', 'mostly_just_before_exams', 'weekend_study_sessions', 'weekly_revision', 'regular_daily_study'],
    'social_media': ['less_than_1_hour', '1-2_hours', '2-4_hours', '4-6_hours', 'more_than_6_hours'],
    'distraction': ['never', 'rarely', 'sometimes', 'often', 'very_often'],
    'assignment_completion': ['never', 'rarely', 'sometimes', 'usually', 'always'],
    'exam_percentage': ['below_40%', '40-49%', '50-59%', '60-69%', '70-79%', '80-89%', '90%_and_above']
}

ordinal_encoder_cols = ['year', 'study_hours', 'sleep_hours', 'attendance',
                        'revision', 'exam_prep', 'social_media', 'distraction',
                        'assignment_completion', 'exam_percentage']

one_hot_encoding_cols = ['education', 'field', 'improvement']

#Split the dictionary into a list of columns and a list of cateogory arrays
columns = list(encoding_dict.keys())
categories = list(encoding_dict.values())

#Initialize ordinal encoder
encoder = OrdinalEncoder(categories = categories, dtype = int)

#Apply to all columns
df_encoded = df
df_encoded[ordinal_encoder_cols] = encoder.fit_transform(df[ordinal_encoder_cols])

#One-hot encode remaining columns
df_encoded = pd.get_dummies(df, one_hot_encoding_cols, drop_first = True, dtype = int)

df_encoded

```

Let's re-add the "improvements" variable, just in-case we want to use it as a target variable in modeling.

```python
df_encoded['improvement'] = df['improvement']

df_encoded
```
